```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# PRECIOS

df = pd.read_csv("../data/raw/precios_spy.csv", sep=";")

plt.figure()
plt.plot(df["Close"])
plt.title("Precio SPY")
plt.savefig("../resultados/precio_spy.png")
plt.show()
```

```python
# PRECIO + SMA

from src.indicadores import add_indicators

df = add_indicators(df)

plt.figure()
plt.plot(df["Close"], label="Precio")
plt.plot(df["SMA_20"], label="SMA 20")
plt.plot(df["SMA_50"], label="SMA 50")
plt.legend()
plt.title("Medias móviles")
plt.savefig("../resultados/sma_spy.png")
plt.show()
```

```python
# SEÑALES

from src.estrategias import estrategia_combinada

df = estrategia_combinada(df)

plt.figure()
plt.plot(df["Close"], label="Precio")
plt.scatter(df.index, df["Close"], c=df["signal"], cmap="coolwarm")
plt.title("Señales de trading")
plt.savefig("../resultados/senales_spy.png")
plt.show()
```

```python
# CURVA DE CAPITAL VS BUY AND HOLD

from src.backtesting import backtest
import numpy as np

df_bt = backtest(df)

# buy & hold
df_bt["buy_hold"] = 10000 * np.exp(df_bt["returns"].cumsum())

plt.figure()
plt.plot(df_bt["equity_curve"], label="Estrategia")
plt.plot(df_bt["buy_hold"], label="Buy & Hold")
plt.legend()
plt.title("Estrategia vs Mercado")
plt.savefig("../resultados/comparacion_spy.png")
plt.show()
```
